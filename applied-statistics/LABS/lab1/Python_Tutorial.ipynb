# List Methods

```python
list1 = ["apple", "banana", "cherry"]
list2 = [1, 5, 7, 9, 3]
list3 = [True, False, False]
list4 = ["abc", 34, True, 40, "male"]

print(len(list4))
```

```python
listx=[] #empyt list
for i in range(5): #i runs from 0 to 4
    x=i+3
    listx.append(x) #adds element to the end
    
print(listx)
```

```python
list_negative=[0]*5 #zeros list
for i in range(1,6): #i runs from 1 to 5
    x=i+2
    list_negative[-i]=x #lists accepts negative indices and counts reversely.
    
print(list_negative)
```

```python
listcopy=listx.copy()
listcopy.append(9)
print(listx)
print(listcopy)
```

```python
listcopy=listx #listcopy and listx refer to same object always!
listcopy.append(9)
print(listx)
print(listcopy)
```

```python
list_unsorted= [4,7,1,14,8,6]
list_increasing=sorted(list_unsorted)
print(list_increasing)
list_decreasing=sorted(list_unsorted,reverse=True)
print(list_decreasing)
```

```python
list_decreasing2= list_increasing[::-1] #reverting a list
print(list_decreasing2)
print(list_increasing)
```

# Defining a Function

```python
def factorial(x):
    product=1
    for i in range(1,x+1):
        product *= i
    return product

print(factorial(5))
```
