```python
#Read the data as different columns
Duration=pd.DataFrame(all["Duration"])
No_Outliers=pd.DataFrame(all["No_Outliers"])

#print(Duration)
#print(No_Outliers)


#Dropping nan from No_Outliers 
NO=No_Outliers.dropna()

print(NO)
```

```python
#Computing the descriptive statistics for Duration

std = np.std(Duration,ddof=1)
mu=np.mean(Duration)
median=np.median(Duration)

mini=np.min(Duration)
maxi=np.max(Duration)
    
Q1=np.quantile(Duration, .25)
Q3=np.quantile(Duration, .75)

skew=Duration.skew(axis = 0, skipna = True) 


#Unformatted output
print(mu, std, mini, maxi, Q1, median, Q3, skew)

```

```python

#Function for all descriptive stats
def descriptive_stats(column):
    sample_mean=np.mean(column)
    sample_std=np.std(column, ddof=1)
    sample_min=np.min(column)
    sample_max=np.max(column)
    quartile_one=np.quantile(column,.25)
    quartile_three=np.quantile(column,.75)
    med_ian= np.median(column)
    skewness=column.skew(axis = 0, skipna = True) 
    
    return sample_mean,sample_std,sample_min, sample_max, quartile_one, med_ian, quartile_three,skewness



#Formatted output
print("    Mean", " Std Dev", "     Min", "     Max", " 1st Quartile", "  Median", " 3rd Quartile", " Skewness")

print("%8.4f %8.4f %8.4f %8.4f %13.4f %8.4f %13.4f %9.4f" % descriptive_stats(Duration))
```

```python
#When the data without outliers is not given 
#Manual way to calculate Duration with NO outliers

# calculate interquartile range
q25, q75 = np.quantile(Duration, .25), np.quantile(Duration, .75)
#print(q25,q75)
iqr = q75 - q25

# calculate whisker cut-off points
cut_off = iqr * 1.5
lower, upper = q25 - cut_off, q75 + cut_off

print(lower,upper)

# identify outliers

npduration=np.array(Duration)
outliers = [x for x in npduration if x < lower or x > upper]
print(outliers)

#remove outliers
outliers_removed = [x for x in npduration if x >= lower and x <= upper]

NOoutlierdata=pd.DataFrame(outliers_removed)
print(NOoutlierdata) 
```
