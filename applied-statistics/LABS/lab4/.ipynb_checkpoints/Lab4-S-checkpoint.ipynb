#### <span style="color:blue">**Descriptive statistics of each column**</span> 

```python
def descriptive_stats(k,column):
    sample_num=k
    sample_mean=np.mean(column)
    sample_std=np.std(column)
    sample_min=np.min(column)
    sample_max=np.max(column)
    quartile_one=np.quantile(column,.25)
    quartile_three=np.quantile(column,.75)
    med_ian= np.median(column)
    skewness=skew(column) 
    
    return sample_num,sample_mean,sample_std,sample_min, sample_max, quartile_one,med_ian, quartile_three,skewness
```

### Illustrating the Central Limit Theorem through simulating exponential random variables

```python
#Generate exponential random variable with rate 0.25 and mean 4 into the matrix (100*10)
NObs=10 #Number of observations in each sample 
rnd_exp=np.random.exponential(scale=4,size=(100,NObs))

# Generating 100 sample means (each mean corresponds to the mean of 10 exponential random variables) - b is the 11th column
for i in rnd_exp:
    smean=np.mean(rnd_exp,axis=1)
smean=np.reshape(smean,(100,1)) # b is the 11th column
rnd_exp=np.append(rnd_exp,smean,axis=1) # we add 11th column in the matrix
```

```python
#Descriptive statistics of 100 exponential random variables (1st and 5th columns) and the means of 10 samples (last column)

print("Column", "    Mean", " Std Dev", "     Min", "     Max", " 1st Quartile", "  Median", " 3rd Quartile", " Skewness")
print("%6d %8.4f %8.4f %8.4f %8.4f %13.4f %8.4f %13.4f %9.4f" % descriptive_stats(1,rnd_exp[:,0]))
print("%6d %8.4f %8.4f %8.4f %8.4f %13.4f %8.4f %13.4f %9.4f" % descriptive_stats(5,rnd_exp[:,4]))
print("%6d %8.4f %8.4f %8.4f %8.4f %13.4f %8.4f %13.4f %9.4f" % descriptive_stats(NObs+1,rnd_exp[:,NObs]))
```

```python
#Boxplots of 100 exponential random variables (1st column) and the means of 10 samples (last column)

plt.figure(figsize=(12, 5)) #figure size for better visibility
plt.subplot(1, 2, 1) 
plt.boxplot(rnd_exp[:,0])
plt.title('Exponential')

plt.subplot(1, 2, 2) 
plt.boxplot(smean)
plt.title('Sample mean')
```

```python
#Histograms of 100 exponential random variables (1st column) and the means of 10 samples (last column)

plt.figure(figsize=(12, 5)) #figure size for better visibility
plt.subplot(1, 2, 1) 
plt.hist(rnd_exp[:,0])
plt.title('Exponential')

plt.subplot(1, 2, 2) 
plt.hist(smean)
plt.title('Sample mean')
```

```python
#Normal probability plots of 100 exponential random variables (1st column) and the means of 10 samples (last column)

exp=rnd_exp[:,0]
ssmean=rnd_exp[:,NObs]

plt.figure(figsize=(12, 5)) #figure size for better visibility
plt.subplot(1, 2, 1) 
probplot(exp, plot=plt) 
plt.title('Exponential vs normal distribution')


plt.subplot(1, 2, 2) 
probplot(ssmean, plot=plt) 
plt.title('Sample mean vs normal distribution')
```

## Illustrating confidence intervals around the mean for normal random variables with known variance 

Let $X$ be a normal random variable with $\mu=5$ and standard deviation $1$. Generate 100 samples of $n=10$ from the random variable $X$.
You can change $n$ value or number of replications (rep) to experiment.

```python
mean1=5
sd1=1
n=10

rep=100

c = np.random.normal(mean1, sd1, size=(rep,n))
for i in c:
    mean_c=np.mean(c,axis=1)
```

Compute the sample mean of each sample.

```python
mean_c=mean_c.reshape((rep,1))   #mean_c is the 11th column
new_a=np.append(c,mean_c,axis=1) #we add 11th column into the matrix
```

Assume that the standard deviation of $X$, $\sigma=1$, is known to you. Compute the confidence intervals for each sample at the confidence levels of $1-\alpha=0.95$.
You can change alpha value here to experiment.

```python
#c_mean=np.mean(mean_c,axis=0) #mean of 11th column

alpha = 0.01
z_score = norm.ppf(1-alpha/2)

#print(z_score)

lcb = mean_c - z_score* (sd1/math.sqrt(n)) #lower limit of the CI

new_a=np.append(new_a, lcb, axis=1) #we add 12th column as lower limit of CI into the matrix


ucb = mean_c + z_score* (sd1/math.sqrt(n)) #upper limit of the CI
new_a=np.append(new_a, ucb, axis=1) #we add 13th column as upper limit of CI into the matrix

#print(new_a)
```

```python
alpha = 0.05
z_score = norm.ppf(1-alpha/2)

print(z_score)
```

Plot the sample means and confidence intervals of the population mean for all samples.

```python
fig = plt.figure(figsize=(20,8))
plt.text(-4.5,np.mean(mean_c)+0.07,"$\mu$", fontsize=15, color = "black", alpha = 0.7) 

for i in range(len(mean_c)):
    horizontal_width = 1
    x_left = i - horizontal_width / 2
    x_right = i + horizontal_width / 2

    ## zorder: The higher the value, more on top the plot shall be.
    plt.plot([i,i], [lcb[i], ucb[i]], color = "dodgerblue",zorder=2) ## line between lower bound and upper bound
    plt.plot([x_left, x_right], [lcb[i], lcb[i]] , color = "dodgerblue",zorder=2) ## horizontal line on lower bound
    plt.plot([x_left, x_right], [ucb[i], ucb[i]] , color = "dodgerblue",zorder=2) ## horizontal line on upper bound
    plt.plot(i, mean_c[i][0], "o", color = "dodgerblue",zorder=2) ## middle point of confidence intervals
    
plt.axhline(np.mean(mean_c), alpha = 0.7, color = "black", ls = "--") ## horizontol line for the mean of all points
plt.show()
```

Write the necessary formulas to count the number of intervals that contain the population mean $5$. Repeat the experiment to observe the changes in the confidence intervals and the number of intervals that contain the population mean $5$.

```python
z = np.ones((rep,1))
new_a=np.append(new_a,z,axis=1)
for i in range(0,rep):
    if 5>new_a[i,n+2] or 5<new_a[i,n+1]:
        new_a[i,n+3]=0
```

How many of the intervals contain the population mean $5$? Interpret your experiment by relating your observations with the definition of confidence intervals.

```python
print(np.sum(new_a[:,n+3],axis=0)) 
```
